The Piotroski score is a discrete score between zero and nine that reflects nine criteria used to determine the strength of a firm's financial position. If a company has a score of 8 or 9, it is considered a good value. If the score adds up to between 0-2 points, the stock is considered weak.

Profitability Criteria Include: 

Positive net income (1 point)

Positive return on assets (ROA) in the current year (1 point)

Positive operating cash flow in the current year (1 point)

Cash flow from operations being greater than net Income (quality of earnings) (1 point)

Leverage, Liquidity, and Source of Funds Criteria Include:

Lower amount of long term debt in the current period, compared to the previous year (decreased leverage) (1 point)

Higher current ratio this year compared to the previous year (more liquidity) (1 point)

No new shares were issued in the last year (lack of dilution) (1 point).

Operating Efficiency Criteria Include:

A higher gross margin compared to the previous year (1 point)

A higher asset turnover ratio compared to the previous year (1 point)

```python
import yfinance as yf #Importing the module

# Giving a list of some of the top companies along with thier tickers so it is easy for the user to enter the stock.
tickerdict={"Amazon" : "AMZN" , "Microsoft" : "MSFT", "APPLE":"AAPL", "TESLA":"TSLA", "ALPHABET(GOOGLE)":"GOOG", "FACEBOOK":"FB","JPMORGAN CHASE":"JPM"}
print("These are some examples of companies with their Tickers(Symbols).")
print(tickerdict)

# Using the stock of Microsoft for this example. 
tick = 'MSFT'
s = yf.Ticker(tick) 
```

```python
# Assigning the value of 0 to the Piotroski score so that we can increase it when the stocks provide positive results based on the indicators.
# Assigning cash flow, finances and balances to thier respective items so it is easier to print the tables in the future
pscore = 0
cashflow = s.cashflow
finances = s.financials
balances = s.balance_sheet
erngs = s.earnings
```

```python
balances
```

```python
cfoa = cashflow.iloc[11,0] #Cashflow from operating activities for the recent quarter
ni = finances.iloc[4,0] #Net Income for the recent quarter
ta = balances.iloc[4,0] #Total Assets as at recent quarter end
ltdc = balances.iloc[23,0] # This is the long term debt of the current year 
ltdp = balances.iloc[23,1]  # This is the long term debt of the previous year 
ca = balances.iloc[18,0] #This is the current assets of the present year
cl = balances.iloc[13,0] #This is the current liabilities of the present year
pca = balances.iloc[18,1] # This this the current assets of the previous year
pcl = balances.iloc[13,1]# This this the current liabilities of the previous year
stockpy = s.cashflow.iloc[6,1] # Stocks issued in the previous year 
stockcy = cashflow.iloc[6,0] # Stocks issued in the current year
tr = finances.iloc[15,0] # Total revenue of the current year 
cr = finances.iloc[17,0] #Cost of revenue for current year 
trpy = finances.iloc[15,1]# Total revenue for previous year 
crpy = finances.iloc[17,1] # Cost of revenue for previous year  
avg_assets = ((balances.iloc[5,0] + balances.iloc[5,1])/2) # Average assets of the current year 
avg_assetspy = ((balances.iloc[5,0] + balances.iloc[5,1])/2)# Average assets of the previous year 
```

**1) NET INCOME :** 
This shows the income of the company for the recent quarter. If the figure is greater than 0 than it indicates that the comapnay is making profits and its performace currently is good impliying that the future performace may also be good and hence a positive indicator if the net income is increasing every year.


```python
if ni>0:
  print("The income of the company for the recent quarter are positive which indicates that the company is not performing poorly")
  pscore+=1
else:
  print("The income of the company for the recent quarter are negative which indicates that the company is performing poorly")
```

**2) RETURN ON ASSETS** :
Return on assets (ROA) is an indicator of how profitable a company is relative to its total assets. ROA gives a manager, investor, or analyst an idea as to how efficient a company's management is at using its assets to generate earnings.

The higher the ROA, the better it is for the company. 

```python
roa = ni/ta #The return on assets ratio
print("The return on assets for the company is: " ,roa)
if roa>0:
  print("The return on assets ratio for the company is positive which implies the company is using its assets efficiently to generate net income.")
  pscore+=1
else:
  print("The return on assets ratio for the company is negative which implies the company is not using its assets effciciently to generate net income.")
```

**3) OPERATING CASH FLOW** : Operating cash flow (OCF) is a measure of the amount of cash generated by a company's normal business operations. Operating cash flow indicates whether a company can generate sufficient positive cash flow to maintain and grow its operations, otherwise, it may require external financing for capital expansion.

```python
if cfoa>0:
  print("The cash flow of the company from operating activities for the recent quarter is positive which indicates that the company is liquid from its operating activities(earns cash)")
  pscore+=1
else:
  print("The cash flow of the company from operating activities for the recent quarter is negative which indicates that the company is not liquid from its operating activities(does not earn cash)")
```

**4) CASH FLOW VS NET INCOME** 
Cash flow refers to the net amount of cash and cash equivalents being transferred in and out of a company. 

Net income (NI), also called net earnings, is calculated as sales minus cost of goods sold, selling, general and administrative expenses, operating expenses, depreciation, interest, taxes, and other expenses

If the cash flow is greater than the net income than it shows that the company has managed to secure good amount of earning and hence it is a positive indicator.

```python
if cfoa>ni:
  print("The Company's Cashflow from Operating Activities for the recent quarter has been greater than its net income for the same period which indicates a good quality of earnings.")
  pscore+=1
else:
  print("The Company's Cashflow from Operating Activities for the recent quarter has been less than its net income for the same period which does not indicate a good quality of earnings and a scope for improvement.")
```

**5) Lower amount of long term debt in the current period, compared to the previous year** : Long-term debt is debt that matures in more than one year. Long-term debt can be viewed from two perspectives: financial statement reporting by the issuer and financial investing.  


If the long term debt has been reducing for many years it indicates financial stability in the company and the comapnies ability to pay its dues time and hence is a positive indicator.

```python
if (ltdp-ltdc) > 0 :
  print("The amount of long term debt has redcued by" ,(ltdp-ltdc), "which indicates the company is working to reduce its debt." )
  pscore+=1
else:
  print(" The amount of long term debt has increased by" ,(-1*(ltdp-ltdc)), "which indicates the comoany has borrowed more recently." )
```

**6) Higher current ratio this year compared to the previous year** : The current ratio is a liquidity ratio that measures a company’s ability to pay short-term obligations or those due within one year. It tells investors and analysts how a company can maximize the current assets on its balance sheet to satisfy its current debt and other payables.

The higher the companies current ratio means that the company is able to manage its assets and liabilities effectively and indicates that the campnay will do well in the short term.


```python
currentratio =  (ca/cl)
print(currentratio)
previousyear_currentratio = (pca/pcl)
print(previousyear_currentratio)
if currentratio > previousyear_currentratio:
  print("Current ratio is higher compared to the previous year which is a good indicator")
  pscore+=1
else:
  print("Current ratio is lower compared to the previous year which is a bad indicator")
```

**7) No new stocks were issued in the last year (lack of dilution)**

```python
newshares =  (stockcy-stockpy)
if newshares <= 0:
  print("No new stocks were issued in the last year.")
  pscore += 1
else:
  print("New stocks were added which is not a good indicator")
```

**8) Gross Margin** :Gross margin is net sales less the cost of goods sold (COGS). In other words, it's the amount of money a company retains after incurring the direct costs associated with producing the goods it sells and the services it provides. 

If the gross margin increases then it is good for the company as it indicates that the comapany is making for profits and might continue to do so. 

```python
grossmargin= tr-cr
print('Gross Margin this year= ' +str(grossmargin))
grossmargin_py = trpy-crpy
print('Gross Margin previous year= ' +str(grossmargin_py))

if grossmargin > grossmargin_py:
  print("The gross margin has increased year on year")
  pscore += 1

else:
  print("The gross margin has decreased year on year.")
```

**9) Asset Turnover Ratio**:The asset turnover ratio measures the value of a company's sales or revenues relative to the value of its assets. The asset turnover ratio can be used as an indicator of the efficiency with which a company is using its assets to generate revenue.

The higher the asset turnover ratio, the more efficient a company is at generating revenue from its assets. Conversely, if a company has a low asset turnover ratio, it indicates it is not efficiently using its assets to generate sales.

```python
asset_turnover_ratio = tr/avg_assets
asset_turnover_ratiopy = trpy/avg_assetspy
print("The asset turnover ratio for this year is: ",asset_turnover_ratio)
print("The asset turnover ratio for the previous year is: ",asset_turnover_ratiopy)

if asset_turnover_ratio>asset_turnover_ratiopy:
  print("The asset turnover ratio has improved which is a good indicator.")
  pscore+=1
else:
  print("The asset turnover ratio has not improved which is a good indicator.")
```

**FINAL RESULT**

```python
print("The piotroski score of the company is: ",pscore)
if pscore > 8:
  print("The company is very strong. You can buy the stock.")
elif pscore<2:
  print("The company is very weak. You should stay away from the stock.")
else:
  print("Buy with caution and proper study.")
```
