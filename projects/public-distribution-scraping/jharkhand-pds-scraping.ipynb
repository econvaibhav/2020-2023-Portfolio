```python
from bs4 import BeautifulSoup
import requests
res = requests.get('https://aahar.jharkhand.gov.in/district-monthly-reports/nfsa-block-reports/MQ%3D%3D/MQ%3D%3D/OA%3D%3D/MQ%3D%3D')
```

```python
res.status_code
```

```python
soup = BeautifulSoup(res.text,'html.parser')
```

```python
table=soup.find_all('table')[1]
```

```python
rows=table.findChildren(['tr'])
```

```python
row1  = []
for r in rows:
    cells = r.find_all(['span','th','td'])
    row1 += [('\t'.join(c.string for c in cells if c.string))]
    #print('\t'.join(c.string for c in cells if c.string))
    #print('00000000000000000')
    #print(row1)
```

```python

import numpy as np 


for j in range(len(row1)):
    row1[j] = row1[j].split('\t')
    for i in range(len(row1[j])):
        row1[j][i] = row1[j][i].replace('\r','')
        row1[j][i] = row1[j][i].replace('\n','')
    
```

```python
import pandas as pd
df2 = pd.DataFrame(row1)
print(df2)
```

```python
from IPython.display import display
#df2.drop(index=df2.index[0])
df2 = df2.drop(2)

#df2.columns = df2.iloc[1]
df2 = df2.drop(1)
df2 = df2.drop(0)
df2.columns = ["Sl No", "Block Name", "HHD Count", "Beneficiary Count (PH Card+AAY Card)","Transaction", "Rice Allocated(kg)", "Wheat Allocated(kg)", "Rice Distributed(kg)","Wheat Distributed(kg)", "Total Allocation", "Total Distributed", "Distribution %"]
display(df2)
```

```python
display(df2.iloc[2])
```

```python
def converter(val):
    if type(val)==str:
        val = val.replace(',','')
    if val == None:
        return float('nan')
    else:
        return float(val)
    
df2['HHD Count'] = df2['HHD Count'].apply(converter)
df2['Beneficiary Count (PH Card+AAY Card)'] = df2['Beneficiary Count (PH Card+AAY Card)'].apply(converter)
df2['Transaction'] = df2['Transaction'].apply(converter)
df2['Rice Allocated(kg)'] = df2['Rice Allocated(kg)'].apply(converter)
df2['Wheat Allocated(kg)'] = df2['Wheat Allocated(kg)'].apply(converter)
df2['Rice Distributed(kg)'] = df2['Rice Distributed(kg)'].apply(converter)
df2['Wheat Distributed(kg)'] = df2['Wheat Distributed(kg)'].apply(converter)
df2['Total Allocation'] = df2['Total Allocation'].apply(converter)
df2['Total Distributed'] = df2['Total Distributed'].apply(converter)
df2['Distribution %'] = df2['Distribution %'].apply(converter)
```
